```python
import pandas as pd
import matplotlib.pyplot as plt

results_path = DATA_DIR / "results/removal_runs_targeted_edge_BE/targeted_removal_seed42_edges1220.csv"
df = pd.read_csv(results_path)

df["num_edges_removed"] = df["step"]
```

```python
plt.figure()
plt.plot(df["num_edges_removed"], df["normalized_efficiency"], marker='o')

plt.xlabel("Number of Edges Removed")
plt.ylabel("Normalized Efficiency")
plt.title("Edges Removed vs Normalized Efficiency")
plt.ylim(0, 1)   # force range from 0 to 1
plt.grid()

plt.show()
```

```python
plt.figure()
plt.plot(
    df["num_edges_removed"],
    df["removal_time_seconds"] / 3600,
    marker='o'
)

plt.xlabel("Number of Edges Removed")
plt.ylabel("Removal Time (hours)")
plt.title("Edges Removed vs Removal Time")
plt.ylim(bottom=0, top=8)   # start y-axis from 0
plt.grid()

plt.show()
```
